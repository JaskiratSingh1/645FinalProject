```python
import pandas as pd

df = pd.read_parquet("hf://datasets/tukey/human_face_emotions_roboflow/data/train-00000-of-00001.parquet")

# Standardize column names (strip whitespace, lower-case, replace spaces with underscores)
df.columns = [col.strip().lower().replace(' ', '_') for col in df.columns]

# Check for missing values
print("Missing values per column:")
print(df.isna().sum())

# No missing values or duplicates, so we can proceed with the data as is

# Print out summary information
print("\nDataframe Info:")
print(df.info())

# Print the first few rows to inspect the data
print("\nFirst 5 rows of the dataset:")
print(df.head())
```

```python
import json

def extract_answer_from_json_str(qa_value):
    # If it's already a list, just return the answer
    if isinstance(qa_value, list):
        return qa_value[0].get("answer", None) if qa_value else None

    # If it's a string, parse it as JSON
    if isinstance(qa_value, str):
        try:
            data = json.loads(qa_value)
            if isinstance(data, list) and len(data) > 0:
                return data[0].get("answer", None)
        except json.JSONDecodeError:
            return None
    
    return None

df["emotion"] = df["qa"].apply(extract_answer_from_json_str)

# Print the first few rows to inspect the data
print("\nFirst 5 rows of the dataset:")
print(df[["qa", "emotion"]].head())
```
